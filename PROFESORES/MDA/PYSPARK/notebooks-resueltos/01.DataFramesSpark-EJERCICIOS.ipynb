1. Crea un DF manual describiendo bebidas (beverages) 🍹

    fabricante
    sabor
    tipo_de_envase
    contenido_de_azúcar_gramos

2. Carga cualquier otro archivo de la carpeta de datos 🏙️

    imprime su esquema
    cuenta el número de filas, llama a `count()`

3. Echa un vistazo a taxi_zones.csv. Lee el archivo a un DF, pero esta vez con tu propio esquema 🎤

```python
from pyspark.sql import Row
from pyspark.sql.types import StructType, StructField, StringType, IntegerType
from pyspark.sql.functions import *


# ============================================================
# 1. Crear un DF manual describiendo bebidas (beverages) 🍹
# ============================================================

bebidas_data = [
    Row(fabricante="Coca-Cola", sabor="Cola", tipo_de_envase="Lata", contenido_de_azúcar_gramos=35),
    Row(fabricante="Pepsi", sabor="Cola", tipo_de_envase="Botella", contenido_de_azúcar_gramos=41),
    Row(fabricante="Fanta", sabor="Naranja", tipo_de_envase="Lata", contenido_de_azúcar_gramos=38),
    Row(fabricante="Nestea", sabor="Té frío", tipo_de_envase="Botella", contenido_de_azúcar_gramos=28),
]

bebidas_df = spark.createDataFrame(bebidas_data)

print("=== DataFrame de bebidas ===")
bebidas_df.show()


# ============================================================
# 2. Cargar otro archivo de la carpeta de datos 🏙️
#    Imprimir esquema + count()
# ============================================================

# Puedes cambiar "books.csv" por cualquier otro archivo disponible
otro_df = (
    spark.read
        .option("header", True)
        .option("inferSchema", True)
        .csv("books.csv")
)

print("=== Esquema del archivo cargado (books.csv) ===")
otro_df.printSchema()

print("=== Número de filas ===")
print(otro_df.count())


# ============================================================
# 3. Leer taxi_zones.csv con un esquema propio 🎤
# ============================================================

schema_taxi = StructType([
    StructField("LocationID", IntegerType(), True),
    StructField("Borough", StringType(), True),
    StructField("Zone", StringType(), True),
    StructField("service_zone", StringType(), True),
])

taxi_df = (
    spark.read
        .option("header", True)
        .schema(schema_taxi)
        .csv("taxi_zones.csv")
)

print("=== Taxi Zones con esquema manual ===")
taxi_df.show(5)
taxi_df.printSchema()
```
