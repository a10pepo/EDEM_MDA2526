1. Prueba como funciona el partition y el coalesce por tu parte con cualquier dataset de los disponibles

```python
# ---------------------------------------
# 1. Leer el CSV
# ---------------------------------------
df = (
    spark.read
        .option("header", "true")
        .option("inferSchema", "true")
        .csv("roles.csv")
)

df.show()
print("Número de particiones iniciales:", df.rdd.getNumPartitions())

# ---------------------------------------
# 2. Demostración de repartition()
# ---------------------------------------

df_rep = df.repartition(4)
print("Particiones después de repartition(4):", df_rep.rdd.getNumPartitions())

# ¿Qué hace repartition?
# - Genera un shuffle completo
# - Redistribuye los datos equilibradamente
# - Útil para paralelizar mejor antes de una operación pesada

# -------------------------------------------------------------------------------------------------------
# 3. Demostración de coalesce() NO CONFUNDAIS CON COALESCE SQL QUE USABAMOS PARA TRABAJAR CON NULLS
# -------------------------------------------------------------------------------------------------------

df_coal = df_rep.coalesce(1)
print("Particiones después de coalesce(1):", df_coal.rdd.getNumPartitions())

# ¿Qué hace coalesce?
# - Reduce particiones sin shuffle (más eficiente)
# - Ideal para escribir en un único archivo de salida

# ---------------------------------------
# 4. Guardar resultado opcional
# ---------------------------------------
# df_coal.write.mode("overwrite").csv("output_unico")
```
