#### Nivel 1 – Bucles
1. Crea una lista con la gente que su nombre tiene 4 letras. 

2. Crea una lista con la gente que su nombre empieza por J y sean menores de 40 años. 

#### Nivel 2 – Funciones

1. Crea una función resta que espere dos parámetros a y b y que devuelva la resta de los mismos. 

2. Crea la función duplicaNumero debe recibir un tipo number y devolver el doble del valor recibido. Si la función no recibe un dato tipo number debe devolver el string ‘Debo ser ejecutada con un número’

3. Crea la función ultimoCaracter debe recibir un tipo string y devolver un string con el último carácter. 

    Si la función no recibe un dato tipo string debe devolver el string 'Debo ser ejecutada con un string'. 

    Si recibe un string vacío debe devolver 'Debo ser ejecutada con un string no vacío' 

4. Crea la función cuentaCaracteres debe recibir un tipo string y devolver un number con el número de carácteres 

    Si la función no recibe un dato tipo string debe devolver el string 'Debo ser ejecutada con un string' 

#### SOLUCIONES PRÁCTICA



```python

gente = [
    {"nombre": "Jamiro", "edad": 45},
    {"nombre": "Juan",   "edad": 35},
    {"nombre": "Paco",   "edad": 34},
    {"nombre": "Pepe",   "edad": 14},
    {"nombre": "Pilar",  "edad": 24},
    {"nombre": "Laura",  "edad": 24},
    {"nombre": "Jenny",  "edad": 10},
]

# =====================================
# NIVEL 1 – BUCLES USANDO append()
# =====================================

# 1. Lista con personas cuyo nombre tiene 4 letras
nombres_4_letras = []  # Lista vacía
for persona in gente:
    if len(persona["nombre"]) == 4:
        nombres_4_letras.append(persona)

print("Nombres con 4 letras:", nombres_4_letras)

# 2. Lista con personas cuyo nombre empieza por 'J' y son menores de 40 años
nombres_j_menores_40 = []  # Lista vacía
for persona in gente:
    if persona["nombre"].startswith("J") and persona["edad"] < 40:
        nombres_j_menores_40.append(persona)

print("Nombres que empiezan por J y menores de 40:", nombres_j_menores_40)

# =====================================
# NIVEL 2 – FUNCIONES
# =====================================

# 1. Función resta
def resta(a, b):
    """
    Devuelve la resta de a - b.
    """
    return a - b

# Ejemplo
print("Resta de 10 - 5:", resta(10, 5))


# 2. Función duplicaNumero
def duplicaNumero(valor):
    """
    Devuelve el doble del número recibido.
    Si no recibe un número, devuelve un mensaje de error.
    """
    if isinstance(valor, (int, float)):
        return valor * 2
    return "Debo ser ejecutada con un número"

# Ejemplo
print("Duplicar 5:", duplicaNumero(5))
print("Duplicar 'texto':", duplicaNumero("texto"))


# 3. Función ultimoCaracter
def ultimoCaracter(cadena):
    """
    Devuelve el último carácter de un string.
    Valida si es string y no está vacío.
    """
    if not isinstance(cadena, str):
        return "Debo ser ejecutada con un string"
    if cadena == "":
        return "Debo ser ejecutada con un string no vacío"
    return cadena[-1]

# Ejemplo
print("Último carácter de 'Hola':", ultimoCaracter("Hola"))
print("Último carácter de cadena vacía:", ultimoCaracter(""))


# 4. Función cuentaCaracteres
def cuentaCaracteres(cadena):
    """
    Devuelve el número de caracteres de un string.
    Si no es string, devuelve un mensaje de error.
    """
    if not isinstance(cadena, str):
        return "Debo ser ejecutada con un string"
    return len(cadena)

# Ejemplo
print("Contar caracteres en 'Python':", cuentaCaracteres("Python"))
print("Contar caracteres en 12345:", cuentaCaracteres(12345))

# =====================================
# RESULTADOS FINALES
# =====================================

print("\n=== RESULTADOS FINALES ===")
print("Personas con nombre de 4 letras:", nombres_4_letras)
print("Personas con nombre que empieza por 'J' y menores de 40:", nombres_j_menores_40)
```
