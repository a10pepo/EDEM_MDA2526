#### Ejercicio 11

#### Ejercicio 12

Escribe una función que reciba un único argumento: un texto (string).
- La función debe devolver un diccionario que contenga el conteo de todas las letras que aparecen en el texto.
- Cada clave del diccionario será una letra, y su valor será la cantidad de veces que esa letra aparece en el texto.

Por ejemplo, para la entrada "Hola Mundo" la función podría devolver un diccionario como:
{'h': 1, 'o': 2, 'l': 1, 'a': 1, 'm': 1, 'u': 1, 'n': 1, 'd': 1} (ignorando mayúsculas/minúsculas).

    TIP: Convierte todo a minúsculas para un conteo uniforme
    TIP 2: Crea un diccionario vacío para guardar el conteo
    TIP 3: debemos ignorar especios y signos, solo contar letras (se te ocurre cómo? recuerda los métodos de strings!)
