## Ejercicio 1
Crea una función que determine qué debe hacer el usuario, según la hora del día. El usuario debe ingresar por consola qué hora es (slo la hora, no los minutos), y el programa le debe decir qué está agendado a esa hora. La casuística es la siguiente:

* Si es entre las 0 y las 8, print "Durmiendo"
* Si es entre las 9 y las 18, print "Trabajando"
* Si es entre las 19 y las 21, print "Clase"
* Si es entre las 22 y las 24, print "Descanso"
* En cualquier otro caso, print "Transporte o error"

Implementa mediante sentencias `if/elif/else`. 

````python
def schedule():
    hora = int(input("Qué hora es? Me vale solo la hora en punto, no los minutos."))
    if hora >= 0 and hora <= 8:
        print("Durmiendo")
    elif hora >= 9 and hora <= 18:
        print("Trabajando")
    elif hora >= 19 and hora <= 21:
        print("Clase")
    elif hora >= 22 and hora <= 24:
        print("Descanso")
    else:
        print("Transporte o error")
schedule()
````

## Ejercicio 4
Escribe una función que le pida al usuario un número del 1 al 100. 

Tienes una variable `lista = [2, 3, 4, 5, 6]`.

El programa debe comprobar que el usuario introduce un número válido. Y luego lo siguiente: 
- Si el número es entre 1 y 30, debe ser multiplicado por la lista de números. El programa debe recorrer la lista e imprimir por pantalla cada elemento de la lista multiplicado por el número uno bajo el otro, separados con este separador: "|"
- Si el número es mayor a 30 y menor a 60, debe ser elevado por esa lista. 
- Si el número es igual o mayor a 60 e igual o menor que 100, debe dividirse por los números de esa lista. 
- Si es mayor a 100, debe lanzar mensaje de error. Esto lo puedes hacer con ifs, o practicar el try/except!

Recuerda documentar las funciones, de a poco ir implementando buenas prácticas!

````python
def procesarNumero():
    """
    Pide un número al usuario (1-100) y realiza diferentes operaciones
    (multiplicación, potencia, división) según el rango del número.
    """
    lista = [2, 3, 4, 5, 6]
    
    try:
        numero = int(input("Dime un número del 1 al 100: "))
    except ValueError:
        print("Error: Por favor, introduce un número entero válido.")
        return

    print("-" * 30)

    # Condición de error: Mayor a 100 o menor a 1
    if numero > 100 or numero < 1:
        print("Error: El número debe estar en el rango del 1 al 100.")
        
    # Condición 1: Multiplicación (1 <= numero <= 30)
    elif 1 <= numero <= 30:
        print(f"Caso: Multiplicación (Número {numero} x Elemento)")
        for elemento in lista:
            resultado = numero * elemento
            print(f"{resultado}|")

    # Condición 2: Potencia (30 < numero < 60)
    elif numero > 30 and numero < 60:
        print(f"Caso: Potencia (Número {numero} ** Elemento)")
        for elemento in lista:
            # Usamos el operador ** para elevar
            resultado = numero ** elemento
            print(f"{resultado}|")

    # Condición 3: División (60 <= numero <= 100)
    elif 60 <= numero <= 100:
        print(f"Caso: División (Número {numero} / Elemento)")
        for elemento in lista:
            # La división siempre resulta en un float (/)
            resultado = numero / elemento
            # Formateamos a 2 decimales para la presentación
            print(f"{resultado:.2f}|") 

# Llamada a la función
procesarNumero()
````

## Ejercicio 9
En este ejercicio vamos a crear un pequeño juego. Se trata de intentar adivinar un numero del 1 al 5. Tenemos dos intentos para acertar. Pasos a seguir:

1. Ya viene implementado cómo obtener un número aleatorio del 1 al 5
2. Tendrás que declarar en una variable el numero de vidas, y mediante un bucle while, comprobar que todavia quedan vidas.
3. Dentro del bucle, obtener el valor del usuario y comprobar si es ese el numero a adivinar. Si no, actualizar las vidas.
4. Si acertamos, salimos del bucle e imprimimos por pantalla "You win". Y si perdemos también salimos del bucle, pero en este caso imprimimos por pantalla "You lose".

TIP: te puede resultar útil usar la sentencia `else` cuando acabe el bucle `while`. Lo que haya dentro de ese `else` se ejecutará una vez acabe la ejecución del `while`. Lo podrás usar para cuando pierdas.

````python
import random

def adivinaElNumero():
    """
    Pequeño juego para adivinar un número del 1 al 5 en dos intentos.
    """

    numero_secreto = random.randint(1, 5)

    vidas = 2
    
    print("¡Bienvenido al juego! Tienes 2 intentos para adivinar un número entre 1 y 5.")

    while vidas > 0:

        print(f"\nTe quedan {vidas} intento(s).")
        
        try:
            # Obtener el valor del usuario
            intento = int(input("Introduce tu número (1-5): "))
        except ValueError:
            print("Entrada inválida. Por favor, introduce un número entero.")
            continue # Vuelve al inicio del bucle sin gastar una vida
            
        # Validación básica del rango
        if not (1 <= intento <= 5):
            print("Número fuera de rango. Debe ser entre 1 y 5.")
            continue # Vuelve al inicio del bucle sin gastar una vida

        # Comprobar si es el número a adivinar
        if intento == numero_secreto:
            # Si acertamos, salimos del bucle e imprimimos "You win"
            print(f"\n¡Felicidades! El número era {numero_secreto}.")
            print("You Has ganado")
            break  # Rompe el bucle while
        else:
            # Si no, actualizamos las vidas
            vidas -= 1
            if vidas > 0:
                print("Incorrecto. ¡Sigue intentándolo!")
                
    # 4. Sentencia else del bucle while (para cuando se pierde)
    # Este bloque 'else' solo se ejecuta si el 'while' termina
    # por la condición (vidas <= 0), no por un 'break'.
    else:
        print(f"\nGame Over. El número secreto era {numero_secreto}.")
        print("Hasta la vista baby")

# Llamada a la función para iniciar el juego
adivinaElNumero()
````

### Ejercicio 13
Escribe un programa que tenga dos variables: un string (pedido al usuario a través de input) y una lista de strings `["Python", "java", "codigo", "ejemplo", "Mundo", "azul"]`. El programa debe recorrer la lista e imprimir por pantalla cada elemento de la lista concatenado con el string, pero: 
- Si la palabra dada por el usuario empieza con vocal, deberás convertirla a MAYÚSCULAS. 
- Caso contrario, debe empezar con mayúsculas y luego minúsculas
Luego comprueba
- Si la palabra tiene más de 4 letras, deberás eliminar las letras "r", "s", "l" o "m" que tenga.   
- Si tiene 4 letras o menos, agrega una carita felis ":)" luego de ésta. 

````python
def procesarTexto():
    """
    Pide un string al usuario, aplica formato y eliminación de caracteres,
    luego concatena el resultado con cada elemento de la lista.
    """
    lista_de_palabras = ["Python", "java", "codigo", "ejemplo", "Mundo", "azul"]
    VOCALES = 'aeiouAEIOU'
    # CORRECCIÓN CLAVE: Incluimos mayúsculas en los caracteres a eliminar
    CARACTERES_A_ELIMINAR = 'rslmRSLM' 

    string_usuario = input("Introduce una palabra: ").strip()

    if not string_usuario:
        print("No se introdujo ninguna palabra. Saliendo del programa.")
        return

    # --- 1. Formateo del String del Usuario ---

    palabra_formateada = ""
    
    if string_usuario[0].lower() in VOCALES.lower():
        # Caso 1: Empieza con vocal -> Convertir a MAYÚSCULAS
        palabra_formateada = string_usuario.upper()
    else:
        # Caso 2: No empieza con vocal -> Primera letra mayúscula, resto minúsculas
        palabra_formateada = string_usuario.capitalize()

    # --- 2. Aplicar eliminación de letras o agregar carita feliz ---
    
    if len(palabra_formateada) > 4:
        # Eliminar las letras "r", "s", "l", "m" (tanto mayúsculas como minúsculas)
        palabra_modificada = palabra_formateada
        for char_a_eliminar in CARACTERES_A_ELIMINAR:
            # Aplicar el reemplazo
            palabra_modificada = palabra_modificada.replace(char_a_eliminar, '')
        
        palabra_final_usuario = palabra_modificada
        
    else:
        # Si tiene 4 letras o menos, agrega una carita felis ":)"
        palabra_final_usuario = palabra_formateada + ":)"

    print("\n--- Resultados del Procesamiento ---")
    print(f"Palabra inicial del usuario: {string_usuario}")
    print(f"Palabra del usuario procesada: {palabra_final_usuario}")
    print("-" * 35)
    
    # --- 3. Recorrer la lista y concatenar ---
    
    for elemento in lista_de_palabras:
        
        # Concatenar el elemento de la lista con la palabra del usuario ya FINALIZADA
        resultado = elemento + " " + palabra_final_usuario
        
        print(resultado)

# Llamada a la función
procesarTexto()
````

````python
def adivinaLaPalabra():
    """
    Juego para adivinar una palabra secreta en dos intentos.
    """

    palabra_secreta = "Python"

    intentos_restantes = 2
    
    print("¡Adivina la Palabra Secreta!")
    print(f"Tienes {intentos_restantes} intentos. (Pista: Es un lenguaje de programación muy popular)")

    while intentos_restantes > 0:
        
        print(f"\nIntento(s) restante(s): {intentos_restantes}")
        

        intento = input("Introduce tu palabra: ").strip() 
        

        if intento.lower() == palabra_secreta.lower():

            print("\n¡Felicidades, has acertado!")
            print("Has ganado")
            break
        else:

            intentos_restantes -= 1
            if intentos_restantes > 0:
                print("Incorrecto. Inténtalo de nuevo.")

    else:

        print("\nGame Over.")
        print(f"Lo siento, has perdido. La palabra secreta era: {palabra_secreta}")

adivinaLaPalabra()
````

## Ejercicio 44
Vamos a crear una clase para representar una biblioteca. La clase debe permitir gestionar libros y su disponibilidad.

* Nombre de la clase: "Biblioteca"
* Atributos comunes:
    * Tipo: "Pública"
    * Horario: "09:00 - 20:00"
* Atributos propios de cada biblioteca:
    * Nombre - String
    * Dirección - String
    * Lista de libros disponibles - Lista de strings
    * Lista de libros prestados - Lista de strings
* Método para añadir un libro a la lista de disponibles
* Método para prestar un libro, que lo elimine de la lista de disponibles y lo añada a la de prestados
* Método para devolver un libro, que lo elimine de la lista de prestados y lo añada a la de disponibles
* Método que imprime el nombre de la biblioteca y su dirección
* Método que devuelve la cantidad total de libros en la biblioteca (sumando disponibles y prestados)

Se pide:
1. Implementar la clase "Biblioteca"
2. Crear dos bibliotecas con datos inventados
3. Añadir libros y prestar algunos de ellos
4. Verificar la cantidad total de libros en ambas bibliotecas
5. Imprimir el nombre y dirección de todas las bibliotecas que contengan la palabra "Central"


````python
class Biblioteca:
    tipo = "Pública"
    horario = "09:00 - 20:00"
    
    def __init__(self, nombre: str, direccion: str):
        self.nombre = nombre
        self.direccion = direccion
        self.libros_disponibles = []
        self.libros_prestados = []
        
    def agregar_libro(self, libro: str):
        self.libros_disponibles.append(libro)
        print(f"'{libro}' añadido a {self.nombre}.")
        
    def prestar_libro(self, libro: str) -> bool:
        """Presta un libro (Disponible -> Prestado)."""
        if libro in self.libros_disponibles:
            self.libros_disponibles.remove(libro)
            self.libros_prestados.append(libro)
            print(f"'{libro}' prestado de {self.nombre}.")
            return True
        else:
            print(f"Error: '{libro}' no está disponible en {self.nombre}.")
            return False
    
    def devolver_libro(self, libro: str) -> bool:
        """Devuelve un libro (Prestado -> Disponible)."""
        if libro in self.libros_prestados:
            self.libros_prestados.remove(libro)
            self.libros_disponibles.append(libro)
            print(f"'{libro}' devuelto a {self.nombre}.")
            return True
        else:
            print(f"Error: '{libro}' no estaba prestado de {self.nombre}.")
            return False
    
    def imprimir_informacion(self):
        """Imprime el nombre de la biblioteca y su dirección."""
        print(f"Biblioteca: {self.nombre}, Dirección: {self.direccion}")
    
    def cantidad_total_libros(self) -> int:
        """Devuelve la cantidad total de libros (disponibles + prestados)."""
        return len(self.libros_disponibles) + len(self.libros_prestados)

# Ejemplo de uso
# 2. Crear dos bibliotecas con datos inventados
biblioteca1 = Biblioteca("Biblioteca Central Madrid", "Avenida del Sol 45")
biblioteca2 = Biblioteca("Biblioteca de Barrio (Gracia)", "Calle Luna 32")
biblioteca3 = Biblioteca("Biblioteca Central de Sevilla", "Plaza Mayor s/n") 

# Lista para manejar las bibliotecas
bibliotecas = [biblioteca1, biblioteca2, biblioteca3]

print("\n--- 3. Añadir libros y prestar algunos de ellos ---")
# Añadir libros a biblioteca 1
biblioteca1.agregar_libro("El Quijote")
biblioteca1.agregar_libro("Cien Años de Soledad")
biblioteca1.agregar_libro("1984")
biblioteca1.prestar_libro("El Quijote")
biblioteca1.prestar_libro("El Quijote") #

# Añadir libros a biblioteca 2
biblioteca2.agregar_libro("Rebelión en la Granja")
biblioteca2.agregar_libro("Orgullo y Prejuicio")
biblioteca2.prestar_libro("Rebelión en la Granja")

# Añadir libros a biblioteca 3
biblioteca3.agregar_libro("La Sombra del Viento")

print("\n--- Verificación de Devolución ---")
biblioteca1.devolver_libro("1984") 
biblioteca1.devolver_libro("El Quijote") 

print("\n--- 4. Verificar la cantidad total de libros ---")
for biblio in bibliotecas:
    print(f"Total de libros en {biblio.nombre}: {biblio.cantidad_total_libros()}")

print("\n--- 5. Imprimir el nombre y dirección de todas las bibliotecas que contengan la palabra 'Central' ---")
for biblio in bibliotecas:
    if "Central" in biblio.nombre:
        biblio.imprimir_informacion()
````

* Método que calcula el precio del alquiler según los kilómetros recorridos, con una tarifa fija de 0.50€/km

Se pide:
1. Implementar la clase "VehículoAlquiler"
2. Crear tres vehículos y alquilar dos de ellos
3. Devolver uno de los vehículos con un kilometraje añadido
4. Imprimir la descripción de todos los vehículos disponibles
5. Calcular el precio del alquiler de uno de los vehículos devueltos


````python
class VehículoAlquiler:

    def __init__(self, marca: str, modelo: str, año: int, kilometraje: float):
        self.marca = marca
        self.modelo = modelo
        self.año = año
        self.kilometraje = kilometraje
        # El estado 'disponible' es un atributo propio de cada instancia.
        self.disponible = True 

    def alquilar(self):
        if self.disponible:
            self.disponible = False
            print(f"{self.obtener_descripcion()} ha sido ALQUILADO.")
        else:
            print(f"Error: {self.obtener_descripcion()} NO está disponible para alquilar.")

    def devolver(self, kilometraje_aumentado: float):
        if not self.disponible:
            self.disponible = True
            self.kilometraje += kilometraje_aumentado
            print(f"{self.obtener_descripcion()} ha sido DEVUELTO. Kilometraje total: {self.kilometraje:.2f} km.")

        else:
            print(f"Error: {self.obtener_descripcion()} ya estaba disponible.")

    def obtener_descripcion(self) -> str:
        return f"{self.marca} {self.modelo} ({self.año})"

    def calcular_precio_alquiler(self, kilometros_recorridos: float) -> float:
        TARIFA = 0.50
        return kilometros_recorridos * TARIFA


print("\n--- 2. Creación y Alquiler ---")
# 2. Crear tres vehículos
vehiculo1 = VehículoAlquiler("Toyota", "Corolla", 2021, 15000)
vehiculo2 = VehículoAlquiler("Ford", "Focus", 2019, 30000)
vehiculo3 = VehículoAlquiler("BMW", "X5", 2022, 5000)

flota = [vehiculo1, vehiculo2, vehiculo3]


vehiculo1.alquilar() 
vehiculo2.alquilar()

print("\n--- 3. Devolver un vehículo ---")

km_recorridos_v1 = 120.5
vehiculo1.devolver(km_recorridos_v1)


print("\n--- 4. Imprimir la descripción de todos los vehículos disponibles ---")
for vehiculo in flota:
    if vehiculo.disponible:
        print(f"DISPONIBLE: {vehiculo.obtener_descripcion()}")

print("\n--- 5. Calcular el precio del alquiler de uno de los vehículos devueltos ---")

precio_v1 = vehiculo1.calcular_precio_alquiler(km_recorridos_v1)
print(f"El precio del alquiler de {vehiculo1.obtener_descripcion()} por {km_recorridos_v1} km es: {precio_v1:.2f}€")
````

## Ejercicio 47
Vamos a crear una clase para modelar una cuenta bancaria.

* Nombre de la clase: "CuentaBancaria"
* Atributos comunes:
    * Moneda: "EUR"
* Atributos propios de cada cuenta:
    * Número de cuenta - String
    * Titular - String
    * Saldo - float
* Método para realizar un depósito, que aumente el saldo con un valor pasado como argumento
* Método para retirar una cantidad, que disminuya el saldo si hay fondos suficientes, sino devuelve un mensaje indicando falta de fondos
* Método para obtener el saldo actual
* Método para transferir dinero a otra cuenta (pasada como argumento), que disminuya el saldo si hay fondos suficientes y aumente el saldo de la cuenta destino

Se pide:
1. Implementar la clase "CuentaBancaria"
2. Crear dos cuentas con información inventada
3. Realizar depósitos y retirar dinero de las cuentas
4. Transferir dinero de una cuenta a otra
5. Imprimir el saldo de ambas cuentas después de la transferencia


````python
class CuentaBancaria:
    # Atributo común (atributo de clase)
    moneda = "EUR"

    def __init__(self, numero_cuenta: str, titular: str, saldo: float):
        self.numero_cuenta = numero_cuenta
        self.titular = titular
        self.saldo = saldo

    def depositar(self, cantidad: float) -> bool:
        """Realiza un depósito y aumenta el saldo."""
        if cantidad > 0:
            self.saldo += cantidad
            print(f"[DEPÓSITO OK] Saldo de {self.titular} (N° {self.numero_cuenta[-4:]}) aumentado en {cantidad:.2f} {self.moneda}.")
            return True
        else:
            print("[ERROR] La cantidad a depositar debe ser positiva.")
            return False

    def retirar(self, cantidad: float) -> bool:
        """Retira una cantidad si hay fondos suficientes."""
        if cantidad > self.saldo:
            print(f"[ERROR RETIRO] Fondos insuficientes para {self.titular}. Saldo actual: {self.saldo:.2f} {self.moneda}.")
            return False
        if cantidad <= 0:
            print("[ERROR RETIRO] La cantidad a retirar debe ser positiva.")
            return False
        
        self.saldo -= cantidad
        print(f"[RETIRO OK] {self.titular} retiró {cantidad:.2f} {self.moneda}. Saldo restante: {self.saldo:.2f} {self.moneda}.")
        return True

    def obtener_saldo(self) -> float:
        """Devuelve el saldo actual."""
        return self.saldo

    def transferir(self, otra_cuenta: 'CuentaBancaria', cantidad: float) -> bool:
        """Transfiere dinero a otra cuenta si hay fondos suficientes."""
        
        if cantidad > self.saldo:
            print(f"[ERROR TRANSFERENCIA] {self.titular} no tiene suficientes fondos para transferir {cantidad:.2f} {self.moneda}.")
            return False
        if cantidad <= 0:
            print("[ERROR TRANSFERENCIA] La cantidad a transferir debe ser positiva.")
            return False

        # 2. Realizar las operaciones
        self.saldo -= cantidad
        otra_cuenta.saldo += cantidad
        
        print(f"[TRANSFERENCIA OK] {self.titular} transfirió {cantidad:.2f} {self.moneda} a {otra_cuenta.titular}.")
        return True


print("--- 2. Creación de Cuentas ---")

cuenta1 = CuentaBancaria("ES1234567890", "Juan Pérez", 1000.00)
cuenta2 = CuentaBancaria("ES0987654321", "María López", 500.00)

print("\n--- 3. Realizar depósitos y retirar dinero ---")

cuenta1.depositar(200.50)
cuenta2.depositar(50.00)


cuenta1.retirar(100.00)
cuenta2.retirar(600.00) # Esto debería fallar

print("\n--- 4. Transferir dinero de una cuenta a otra ---")

monto_transferencia = 300.00
cuenta1.transferir(cuenta2, monto_transferencia)

cuenta1.transferir(cuenta2, 900.00)

print("\n--- 5. Imprimir el saldo de ambas cuentas después de la transferencia ---")
print(f"Saldo final de {cuenta1.titular}: {cuenta1.obtener_saldo():.2f} {cuenta1.moneda}") 
print(f"Saldo final de {cuenta2.titular}: {cuenta2.obtener_saldo():.2f} {cuenta2.moneda}")
````

* Método para dar una proyección de las ventas en caso de haber invertido X dinero en marketing en esa tienda. Siendo X el parámetro de entrada. Si X es menor de 1000, las ventas de los ultimos 3 meses hubiesen sido de (1.2 x ventas), si es mayor o igual de 1000, las ventas hubiesen sido de (1.5 x venta). El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve las ventas con el aumento.

Se pide:
1. Implementar la clase "Tienda"

3. Comprueba en al menos una de ellas, todo lo implementado en la clase tienda (sus atributos, media de ventas, ventas/empleado...)

4. Imprime por pantalla los nombres de las tiendas cuya dirección lleve el string "Avenida"

````python
class Tienda:
    """
    Representa una tienda de electrodomésticos con sus métricas de ventas.
    """

    tipo = "Electrodomésticos"
    abierta = True

    def __init__(self, nombre: str, direccion: str, num_empleados: int, ventas_3m: list):
        """
        Inicializa una nueva instancia de Tienda.

        Args:
            nombre (str): Nombre de la tienda.
            direccion (str): Dirección física de la tienda.
            num_empleados (int): Número de empleados.
            ventas_3m (list): Lista de 3 floats/ints con las ventas de los últimos 3 meses.
        """
        self.nombre = nombre
        self.direccion = direccion
        self.num_empleados = num_empleados
        self.ventas_3m = ventas_3m

    def ventas_totales(self) -> float:
        """Calcula y devuelve la suma total de las ventas de los últimos 3 meses."""
        return sum(self.ventas_3m)
    
    def media_ventas_por_empleado(self) -> float:
        """Calcula la media de ventas totales por empleado."""
        if self.num_empleados == 0:
            return 0.0  # Evita la ZeroDivisionError
        return self.ventas_totales() / self.num_empleados
    
    def obtener_resumen(self) -> str:
        """Devuelve un string con el nombre y la dirección de la tienda."""
        return f"Tienda: {self.nombre} | Dirección: {self.direccion}"
    
    def ventas_ultimo_mes(self) -> float:
        """Obtiene las ventas del último mes (último elemento de la lista)."""
        if not self.ventas_3m:
            return 0.0
        return self.ventas_3m[-1]
        
    def proyectar_ventas(self, inversion_marketing: float) -> list:
        """
        Proyecta y actualiza las ventas de los últimos 3 meses basándose en una inversión en marketing.
        
        Args:
            inversion_marketing (float): Cantidad invertida en marketing.

        Returns:
            list: Lista de las ventas proyectadas (actualizado como atributo).
        """
        if inversion_marketing < 1000:
            incremento = 1.2
        else:
            incremento = 1.5

        ventas_proyectadas = [venta * incremento for venta in self.ventas_3m]
        # LO MISMO SERÍA:
        # ventas_proyectadas = []  # Inicializas una lista vacía
        # for venta in self.ventas_3m:
        #     resultado = venta * incremento
        #     ventas_proyectadas.append(resultado)
        
        # Reescribe el atributo
        self.ventas_3m = ventas_proyectadas
        return self.ventas_3m


print("--- 2. Creación de Tres Tiendas ---")

tienda_a = Tienda("Central Chamberí", "Calle Alcalá 25", 5, [15000.0, 18000.0, 22000.0])
tienda_b = Tienda("Outlet Las Rosas", "Avenida de América 100", 3, [8000.0, 9500.0, 7000.0])
tienda_c = Tienda("Tienda Moncloa", "Avenida Principal 33", 4, [12000.0, 14000.0, 16000.0])

flota_tiendas = [tienda_a, tienda_b, tienda_c]


print("\n--- 3. Comprobar todos los métodos de una tienda (Tienda A) ---")
print(f"Nombre y Dirección: {tienda_a.obtener_resumen()}")
print(f"Tipo (Común): {tienda_a.tipo}")
print(f"Total Ventas: {tienda_a.ventas_totales():.2f}€")
print(f"Media/Empleado: {tienda_a.media_ventas_por_empleado():.2f}€")
print(f"Ventas Último Mes: {tienda_a.ventas_ultimo_mes():.2f}€")

inversion = 2500
print(f"\nProyección de Ventas (Inversión: {inversion}€):")
ventas_proy = tienda_a.proyectar_ventas(inversion)
print(f"Ventas proyectadas: {ventas_proy}")
print(f"Nuevas Ventas Totales: {tienda_a.ventas_totales():.2f}€ (Atributo actualizado)")


print("\n--- 4. Calcular las ventas del último mes de todas las tiendas ---")
for tienda in flota_tiendas:
    print(f"Ventas de '{tienda.nombre}' (Último Mes): {tienda.ventas_ultimo_mes():.2f}€")


print("\n--- 5. Imprime nombres de tiendas cuya dirección lleve el string 'Avenida' ---")
for tienda in flota_tiendas:
    if "avenida" in tienda.direccion.lower(): 
        print(f"Tienda con 'Avenida': {tienda.nombre}")

print("\n--- 5. Documentación de la clase ---")
print("La documentación (docstrings) ha sido incluida dentro de la definición de la clase y sus métodos.")
````

## Ejercicio 49
En este ejercicio vamos a implementar una clase *Perro* en Python. La clase tiene las siguientes características:
* Cosas que sabemos seguro que tiene un perro
    * Tiene 4 patas
    * 2 orejas
    * 2 ojos
    * Una velocidad de 0. Por defecto, el perro está parado
* Cuando se inicialice:
    * El perro será de una determinada raza
    * Por defecto tendrá pelo "Marrón", a no ser que se diga lo contrario.
    * Por defecto no tendrá dueño, a no ser que se diga lo contrario.
    
* Dispondrá también de un método llamado andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro.
* Necesita otro método (parar), donde pondremos la velocidad a 0.
* Otro método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada.


Se pide:
* Implementa la clase *Perro*
* Crea un objeto de tipo *Perro*, sin dueño
* Comprueba que están bien todos sus atributos
* Prueba que ande, y comprueba su velocidad
* Páralo
* Documenta la clase *Perro*

````python
class Perro:
    patas = 4
    orejas = 2
    ojos = 2

    def __init__(self, raza: str, pelo: str = "Marrón", dueño: bool = False):
        """
        Inicializa un nuevo perro.

        Args:
            raza (str): La raza del perro.
            pelo (str, optional): El color del pelo. Por defecto, "Marrón".
            dueño (bool, optional): Indica si el perro tiene dueño. Por defecto, False.
        """
        self.pelo = pelo
        # Se renombra 'owner' a 'dueño' para mantener consistencia con el enunciado
        self.dueño = dueño 
        self.raza = raza
        # Velocidad inicial propia de este perro (Atributo de Instancia)
        self.velocidad = 0 

    def andar(self, aumento_velocidad: float):
        """
        Aumenta la velocidad del perro.

        Args:
            aumento_velocidad (float): El valor a sumar a la velocidad actual.
        """
        self.velocidad += aumento_velocidad

    def parar(self):
        """
        Pone la velocidad del perro a cero.
        """
        self.velocidad = 0

    def ladrar(self, mensaje: str) -> str:
        """
        Devuelve el sonido del ladrido concatenado con un mensaje.

        Args:
            mensaje (str): El argumento de entrada a concatenar.
            
        Returns:
            str: El ladrido con el mensaje.
        """
        # Se ajusta la salida para que coincida con el formato del enunciado
        return f"GUAU! {mensaje}"
````
