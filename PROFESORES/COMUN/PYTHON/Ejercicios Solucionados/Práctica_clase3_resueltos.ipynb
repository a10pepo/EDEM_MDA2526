#### Ejercicio 1
Crea una función que convierta un str en un número. Si el argumento pasado no es convertible, debe imprimir un mensaje de error

```python
def convertir_a_numero(cadena):
    if cadena.isdigit():
        return int(cadena)
    else:
        print(f"Error: '{cadena}' no es un número entero positivo válido.")
        return None

print(convertir_a_numero("123"))   # 123
print(convertir_a_numero("-123"))  # Error y None
print(convertir_a_numero("3.14"))  # Error y None
print(convertir_a_numero("abc"))   # Error y None
```

#### Ejercicio 2

Crea una función llamada calcular_promedio que:

- Reciba una lista de números (enteros o flotantes).

- Calcule y devuelva el promedio (media aritmética) de los números en la lista.

Si la lista está vacía, debe devolver None y mostrar un mensaje indicando que no hay datos para calcular.

```python
def calcular_promedio(lista_numeros):
    if len(lista_numeros) == 0:
        print("Lista vacía, no se puede calcular promedio")
        return None
    suma = sum(lista_numeros)
    return suma / len(lista_numeros)


print(calcular_promedio([10, 20, 30]))  # 20.0
print(calcular_promedio([]))             # None y mensaje
```

#### Ejercicio 3

Crea una función restar_numeros_pares(*args) que resten a la tabliable predefinida "total=50" todos los números pares que se le pasen como argumentos posicionales. 

```python
def restar_numeros_pares(*args):
    total = 50
    for num in args:
        if num % 2 == 0:
            total -= num
    return total

print(restar_numeros_pares(1, 2, 3, 4, 5))
```

#### Ejercicio 4

Define una función imprimir_perfil(**kwargs) que reciba un diccionario y muestre la información como "clave: valor" para cada elemento. 

```python
def imprimir_perfil(**kwargs):
    for clave, valor in kwargs.items():
        print(f"{clave}: {valor}")

datos_perfil = {"nombre": "Ana", "edad": 28, "ciudad": "Bogotá"}
imprimir_perfil(**datos_perfil)
```

#### Ejercicio 9
Crea un programa que recoja dos inputs. Tiene que comprobar si su suma es igual, superior o inferior a 10.

```python
numero_1 = int(input("Elije un numero"))
numero_2 = int(input("Elige otro numero"))
if (numero_1 + numero_2) == 10:
    print((numero_1+numero_2),"Es igual que 10")
elif (numero_1 + numero_2) > 10:
    print((numero_1+numero_2),"Es mayor que 10")
elif (numero_1 + numero_2) < 10:
    print((numero_1+numero_2),"Es menor que 10")
```

#### Ejercicio 10

- Haz una lista con números enteros. 
- Inicializa una variable "total" en 0. 
- Escribe un programa que calcule la suma acumulada hasta que el total exceda 100. 
- Imprime cada valor que se suma y cuándo el total se pase de 100, imprime "Límite superado" y termina.

```python
lista = [1, 15, 7, 8, 5, 3, 0, 11, 4, 13]
total = 0 
for num in lista:
    total += num
    print(f"Sumando {num}, total = {total}")
    if total > 100:
        print("Límite superado")
        break
```

#### Ejercicio 11
Tienes la siguiente lista:

> list1 = [12, 15, 32, 42, 55, 75, 122, 132, 150, 180, 200]

Implementa un programa que recorra cada elemento e imprima por pantalla todos los que sean divisibles por 5. Si nos encontramos con alguno que sea mayor que 150, detener el bucle.

```python
list1 = [12, 15, 32, 42, 55, 75, 122, 132, 150, 180, 200]
for num in list1:
    if num > 150:
        break
    if num % 5 == 0:
        print(num)
```

#### Ejercicio 12

Crea un programa que:
- Pida al usuario ingresar un precio con decimales (por ejemplo, 12.98765).
- Pida al usuario cuántos decimales quiere mostrar (un número entero).
- Use la función round() para redondear el precio al número de decimales indicados.
- Imprima el precio redondeado con el formato:
"El precio redondeado es: [precio]"

```python
# Pedir al usuario que ingrese un precio con decimales
precio = float(input("Ingresa el precio: "))

# Pedir al usuario cuántos decimales quiere mostrar
decimales = int(input("¿Cuántos decimales quieres mostrar? "))

# Redondear el precio con la función round()
precio_redondeado = round(precio, decimales)

# Mostrar el resultado
print("El precio redondeado es:", precio_redondeado)
```
