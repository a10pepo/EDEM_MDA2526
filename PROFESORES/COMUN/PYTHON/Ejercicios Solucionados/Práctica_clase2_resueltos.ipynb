#### Ejercicio 6

Escribe un programa que simule una cuenta atrás desde un número n dado (que se pasará como argumento al llamar la función). 
- En cada pase del bucle, imprime el número. 
- PERO: si el número es divisible por 4, en vez de imprimir el número, imprime "Pum!". 
- Cuando llegue a 0, imprime "¡Despegue!".

```python
def cuenta_atras(n):
    for i in range(n, -1, -1):
        if i % 4 == 0 and i != 0:
            print("Pum!")
        elif i == 0:
            print("¡Despegue!")
        else:
            print(i)

cuenta_atras(15)
```

#### Ejercicio 7
Para que no se olvide, repasemos con este ejercicio los métodos de strings (puedes revisarlos [aquí](https://www.w3schools.com/python/python_strings_methods.asp)).

Dada esta cadena de texto: `"Ole Python como mola! Me está encantando. De verdad que SÍ"`
- Cuenta cuántas letras están en mayúscula, cuántas en minúscula y cuántos dígitos hay en total. 
- Luego imprime esos tres totales, usando el formato `(f"{}")`

```python
texto = "Ole Python mola 1 montón! Me ENCANTA!"

mayusculas = 0
minusculas = 0
digitos = 0

for c in texto:
    if c.isupper():
        mayusculas += 1
    elif c.islower():
        minusculas += 1
    elif c.isdigit():
        digitos += 1

print(f"Mayúsculas: {mayusculas}")
print(f"Minúsculas: {minusculas}")
print(f"Dígitos: {digitos}")
```

Escribe una función que convierta números del 1 al 7 en el nombre del día de la semana correspondiente (como string). La función constará de un único argumento numérico y una salida de tipo string. 
- Una pista: puedes crear un diccionario con {clave-número: valor-"Día"}
- Otra opción es usar condicionales y operadores de igualdad.

```python
def dia_de_la_semana(numero):
    dias = {
        1: "Lunes",
        2: "Martes",
        3: "Miércoles",
        4: "Jueves",
        5: "Viernes",
        6: "Sábado",
        7: "Domingo"
    }
    return dias.get(numero, "Número fuera de rango")

# Ejemplos de uso
print(dia_de_la_semana(1))  # Output: Lunes
print(dia_de_la_semana(5))  # Output: Viernes
print(dia_de_la_semana(8))  # Output: Número fuera de rango
```

```python
def dia_de_la_semana(numero):
    if numero == 1:
        return "Lunes"
    elif numero == 2:
        return "Martes"
    elif numero == 3:
        return "Miércoles"
    elif numero == 4:
        return "Jueves"
    elif numero == 5:
        return "Viernes"
    elif numero == 6:
        return "Sábado"
    elif numero == 7:
        return "Domingo"
    else:
        return "Número fuera de rango"

# Ejemplo de uso
print(dia_de_la_semana(3))  # Miércoles
print(dia_de_la_semana(8))  # Número fuera de rango
```

Escribe una función que reciba dos argumentos:

#### Ejercicio 12

Escribe una función que reciba un único argumento: un texto (string).
- La función debe devolver un diccionario que contenga el conteo de todas las letras que aparecen en el texto.
- Cada clave del diccionario será una letra, y su valor será la cantidad de veces que esa letra aparece en el texto.

Por ejemplo, para la entrada "Hola Mundo" la función podría devolver un diccionario como:
{'h': 1, 'o': 2, 'l': 1, 'a': 1, 'm': 1, 'u': 1, 'n': 1, 'd': 1} (ignorando mayúsculas/minúsculas).

    TIP: Convierte todo a minúsculas para un conteo uniforme
    TIP 2: Crea un diccionario vacío para guardar el conteo
    TIP 3: debemos ignorar especios y signos, solo contar letras (se te ocurre cómo? recuerda los métodos de strings!)

```python
def contar_letras(texto):
    texto = texto.lower()        
    conteo = {}               

    for letra in texto:
        if letra.isalpha():      
            if letra in conteo:
                conteo[letra] += 1
            else:
                conteo[letra] = 1

    return conteo

resultado = contar_letras("Estos ejercicios están geniales, de 10!")
print(resultado)
```

#### Ejercicio 16

Crea una función que reciba un número variable de argumentos posicionales usando `*args`, todos números.

La función debe devolver la suma de todos esos números.

Ejemplo: sumar_todos(1, 3, 5, 7) debe devolver 16.

```python
def sumar_todos(*args):
    total = 0
    for numero in args:
        total += numero
    return total

resultado = sumar_todos(1, 3, 5, 7)
print(resultado) 
```

Define una función que reciba argumentos de palabra clave (**kwargs), que sea información de un socio de una biblioteca.

La función debe imprimir cada clave y su valor, uno por línea, de forma clara.
