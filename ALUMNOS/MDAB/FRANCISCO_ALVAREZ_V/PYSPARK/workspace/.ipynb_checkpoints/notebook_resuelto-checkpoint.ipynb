# EJERCICIO FINAL PYSPARK (DOCKER + KAFKA + PYSPARK)
## Solución Completa

````python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql import functions as F
from pyspark.sql.window import Window
import matplotlib.pyplot as plt
import pandas as pd
````

## Creamos la sesión de spark

````python
spark = (
    SparkSession.builder
        .appName("ProyectoIoT")
        .getOrCreate()
)
spark
````

## Configuración del Streaming desde Kafka

spark.readStream indica que vamos a leer datos en modo streaming, es decir, datos que llegan continuamente.

.format("kafka") le decimos a Spark que la fuente de datos de streaming será Kafka.

.option("kafka.bootstrap.servers", "kafka:9092") especifica la dirección del clúster Kafka al que conectarse.
En este caso "kafka" es el nombre del contenedor Docker y 9092 es el puerto del broker.

.option("subscribe", "sensores") indica el topic de Kafka del que queremos leer mensajes → "sensores".

.option("startingOffsets", "latest") define desde qué punto empezar a leer:
- "latest" → solo recibe mensajes nuevos, enviados después de iniciar el streaming.
- "earliest" sería para leer todos los mensajes antiguos también.

.load() ejecuta la configuración y crea un DataFrame de streaming.

````python
raw_df = (
    spark.readStream
         .format("kafka")
         .option("kafka.bootstrap.servers", "kafka:9092")
         .option("subscribe", "sensores")
         .option("startingOffsets", "latest")
         .load()
)
````

## Transformación de los mensajes JSON

Este bloque transforma los mensajes de Kafka en un formato que Spark puede procesar:

- Define la estructura de los datos que esperamos recibir (sensor, valor, temperatura, humedad, estado, timestamp y UUID).
- Convierte el JSON recibido en columnas individuales para poder trabajar con cada campo.
- Crea una columna de tiempo (event_time) a partir del timestamp original para poder usarlo en ventanas y agregaciones temporales.

````python
schema = StructType([
    StructField("sensor_id", StringType()),
    StructField("value", DoubleType()),
    StructField("temperature", DoubleType()),
    StructField("humidity", DoubleType()),
    StructField("status", StringType()),
    StructField("timestamp", DoubleType()),
    StructField("uuid", StringType())
])

df = raw_df.select(
    from_json(col("value").cast("string"), schema).alias("data")
).select("data.*")

df = df.withColumn("event_time", col("timestamp").cast("timestamp"))
````

## Agregaciones por ventana de tiempo

Este bloque realiza agregaciones por ventana de tiempo sobre el DataFrame de streaming:

- Define un watermark de 1 minuto para indicar a Spark hasta qué punto los datos antiguos se pueden considerar válidos.
- Agrupa los datos cada 30 segundos por sensor (sensor_id).
- Calcula estadísticas dentro de cada ventana: promedio de valor, temperatura y humedad, y cuenta el número de eventos recibidos.

````python
agg_df = (
    df
    .withWatermark("event_time", "1 minute")
    .groupBy(
        window(col("event_time"), "30 seconds"),
        col("sensor_id")
    )
    .agg(
        avg("value").alias("avg_value"),
        avg("temperature").alias("avg_temp"),
        avg("humidity").alias("avg_humidity"),
        count("*").alias("num_events")
    )
)
````

## Escritura en archivos Parquet

Este bloque escribe los resultados del streaming en archivos Parquet de manera continua:

- Define la carpeta de salida donde se guardarán los archivos Parquet (resultados/).
- Usa un checkpoint (chk/) para que Spark recuerde qué datos ya procesó y pueda reiniciar de forma segura en caso de fallo.
- Modo append: los nuevos datos se agregan continuamente a los archivos existentes.

````python
parquet_query = (
    agg_df
    .writeStream
    .format("parquet")
    .option("path", "resultados/")
    .option("checkpointLocation", "chk/")
    .outputMode("append")
    .start()
)
````

### Ejercicio 4: Agregaciones por sensor
Agrupa el DataFrame por sensor_id y obtén estadísticas descriptivas.

Responde:
- ¿Qué sensor tiene la mayor media en la variable analizada?
- ¿Qué sensor presenta la temperatura máxima?
- ¿Cuántos grupos o ventanas existen por sensor?

````python
# Agregaciones por sensor
stats_by_sensor = df_agg.groupBy("sensor_id").agg(
    F.avg("avg_value").alias("media_valor"),
    F.max("avg_value").alias("max_valor"),
    F.min("avg_value").alias("min_valor"),
    F.avg("avg_temp").alias("media_temp"),
    F.max("avg_temp").alias("max_temp"),
    F.min("avg_temp").alias("min_temp"),
    F.avg("avg_humidity").alias("media_humedad"),
    F.sum("num_events").alias("total_eventos"),
    F.count("*").alias("num_ventanas")
).orderBy("sensor_id")

print("=== ESTADÍSTICAS POR SENSOR ===")
stats_by_sensor.show(truncate=False)

# Sensor con mayor media en avg_value
sensor_max_media = stats_by_sensor.orderBy(col("media_valor").desc()).first()
print(f"\n=== RESPUESTAS ===")
print(f"Sensor con mayor media en avg_value: {sensor_max_media['sensor_id']} (media: {sensor_max_media['media_valor']:.2f})")

# Sensor con temperatura máxima
sensor_max_temp = stats_by_sensor.orderBy(col("max_temp").desc()).first()
print(f"Sensor con temperatura máxima: {sensor_max_temp['sensor_id']} (temperatura: {sensor_max_temp['max_temp']:.2f}°)")

# Ventanas por sensor
print(f"\n=== NÚMERO DE VENTANAS (GRUPOS) POR SENSOR ===")
stats_by_sensor.select("sensor_id", "num_ventanas").show()
````

### Ejercicio 5: Ranking con funciones de ventana
Usa funciones de ventana para identificar, para cada sensor, el registro con el mayor valor de avg_value.

Responde:
- ¿Qué sensor obtiene el valor máximo global entre todos?

````python
# Definir una ventana particionada por sensor_id, ordenada por avg_value descendente
window_spec = Window.partitionBy("sensor_id").orderBy(col("avg_value").desc())

# Agregar columna de ranking
df_ranked = df_agg.withColumn("rank", F.rank().over(window_spec))

# Filtrar solo los registros con rank = 1 (máximo valor por sensor)
df_max_by_sensor = df_ranked.filter(col("rank") == 1)

print("=== REGISTRO CON MAYOR avg_value POR SENSOR ===")
df_max_by_sensor.select(
    "sensor_id", "window", "avg_value", "avg_temp", "avg_humidity", "num_events"
).orderBy("sensor_id").show(truncate=False)

# Encontrar el sensor con el valor máximo global
max_global = df_max_by_sensor.orderBy(col("avg_value").desc()).first()

print(f"\n=== VALOR MÁXIMO GLOBAL ===")
print(f"Sensor: {max_global['sensor_id']}")
print(f"Valor máximo: {max_global['avg_value']:.2f}")
print(f"Temperatura en ese momento: {max_global['avg_temp']:.2f}°")
print(f"Humedad en ese momento: {max_global['avg_humidity']:.2f}%")
print(f"Ventana temporal: {max_global['window']['start']} a {max_global['window']['end']}")
````

Responde:
- ¿Cuál es la distribución de registros entre las particiones?

### Ejercicio 9: Detección de anomalías
Define un criterio personalizado de "anomalía" basado en los valores medios del DataFrame.

Responde:
- ¿Cuántos registros son anómalos?
- ¿Qué sensor tiene más anomalías?

````python
# Calcular estadísticas globales para definir umbrales de anomalía
stats_globales = df_agg.select(
    F.avg("avg_value").alias("media_valor"),
    F.stddev("avg_value").alias("stddev_valor"),
    F.avg("avg_temp").alias("media_temp"),
    F.stddev("avg_temp").alias("stddev_temp"),
    F.avg("avg_humidity").alias("media_humidity"),
    F.stddev("avg_humidity").alias("stddev_humidity")
).collect()[0]

media_valor = stats_globales['media_valor']
stddev_valor = stats_globales['stddev_valor']
media_temp = stats_globales['media_temp']
stddev_temp = stats_globales['stddev_temp']
media_humidity = stats_globales['media_humidity']
stddev_humidity = stats_globales['stddev_humidity']

print("=== ESTADÍSTICAS GLOBALES PARA DETECCIÓN DE ANOMALÍAS ===")
print(f"Valor medio: {media_valor:.2f} ± {stddev_valor:.2f}")
print(f"Temperatura media: {media_temp:.2f} ± {stddev_temp:.2f}")
print(f"Humedad media: {media_humidity:.2f} ± {stddev_humidity:.2f}")

# Definir criterios de anomalía:
# 1. Valores fuera de 2 desviaciones estándar
# 2. Temperatura muy alta (>80) o muy baja (<30)
# 3. Humedad extrema (<25 o >85)
# 4. Número de eventos muy bajo (<15)

df_anomalias = df_agg.withColumn(
    "es_anomalia",
    F.when(
        (F.abs(col("avg_value") - media_valor) > 2 * stddev_valor) |
        (col("avg_temp") > 80) |
        (col("avg_temp") < 30) |
        (col("avg_humidity") < 25) |
        (col("avg_humidity") > 85) |
        (col("num_events") < 15),
        True
    ).otherwise(False)
)

# Crear columna con tipo de anomalía
df_anomalias = df_anomalias.withColumn(
    "tipo_anomalia",
    F.when(F.abs(col("avg_value") - media_valor) > 2 * stddev_valor, "valor_extremo")
     .when((col("avg_temp") > 80) | (col("avg_temp") < 30), "temperatura_extrema")
     .when((col("avg_humidity") < 25) | (col("avg_humidity") > 85), "humedad_extrema")
     .when(col("num_events") < 15, "pocos_eventos")
     .otherwise("normal")
)

# Contar anomalías
num_anomalias = df_anomalias.filter(col("es_anomalia") == True).count()
total_registros = df_anomalias.count()
porcentaje_anomalias = (num_anomalias / total_registros) * 100

print(f"\n=== RESULTADOS DE DETECCIÓN DE ANOMALÍAS ===")
print(f"Total de registros anómalos: {num_anomalias}")
print(f"Porcentaje de anomalías: {porcentaje_anomalias:.2f}%")

# Contar anomalías por sensor
anomalias_por_sensor = df_anomalias.filter(col("es_anomalia") == True).groupBy("sensor_id").agg(
    F.count("*").alias("num_anomalias")
).orderBy(col("num_anomalias").desc())

print("\n=== ANOMALÍAS POR SENSOR ===")
anomalias_por_sensor.show()

sensor_mas_anomalias = anomalias_por_sensor.first()
print(f"\nSensor con más anomalías: {sensor_mas_anomalias['sensor_id']} ({sensor_mas_anomalias['num_anomalias']} anomalías)")

# Tipos de anomalías
print("\n=== DISTRIBUCIÓN POR TIPO DE ANOMALÍA ===")
df_anomalias.groupBy("tipo_anomalia").count().orderBy(col("count").desc()).show()

# Mostrar ejemplos de anomalías
print("\n=== EJEMPLOS DE REGISTROS ANÓMALOS ===")
df_anomalias.filter(col("es_anomalia") == True).select(
    "sensor_id", "avg_value", "avg_temp", "avg_humidity", "num_events", "tipo_anomalia"
).show(10, truncate=False)
````

### Ejercicio 10: Cruce de métricas
Crea una puntuación combinando varias columnas del DataFrame (por ejemplo, ponderando avg_value, temperatura y humedad).

Responde:
- ¿Qué sensor obtiene la puntuación más alta?
- ¿Cuál es el valor de dicha puntuación?

````python
# Crear puntuación combinada con ponderaciones
# Fórmula: score = (avg_value * 0.4) + (avg_temp * 0.3) + (avg_humidity * 0.2) + (num_events * 0.1)
# Esta fórmula da más peso al valor del sensor, seguido de temperatura, humedad y eventos

df_score = df_agg.withColumn(
    "score_total",
    (col("avg_value") * 0.4) + 
    (col("avg_temp") * 0.3) + 
    (col("avg_humidity") * 0.2) + 
    (col("num_events") * 0.1)
)

# También podemos crear un score de "eficiencia" (valor alto con temperatura y humedad moderadas)
df_score = df_score.withColumn(
    "score_eficiencia",
    col("avg_value") * (1 - (F.abs(col("avg_temp") - 55) / 100)) * (1 - (F.abs(col("avg_humidity") - 55) / 100))
)

# Encontrar el registro con mayor puntuación total
max_score_registro = df_score.orderBy(col("score_total").desc()).first()

print("=== PUNTUACIÓN COMBINADA ===")
print("Fórmula: (avg_value × 0.4) + (avg_temp × 0.3) + (avg_humidity × 0.2) + (num_events × 0.1)")
print(f"\nRegistro con mayor puntuación total:")
print(f"  Sensor: {max_score_registro['sensor_id']}")
print(f"  Puntuación: {max_score_registro['score_total']:.2f}")
print(f"  Valor promedio: {max_score_registro['avg_value']:.2f}")
print(f"  Temperatura: {max_score_registro['avg_temp']:.2f}°")
print(f"  Humedad: {max_score_registro['avg_humidity']:.2f}%")
print(f"  Eventos: {max_score_registro['num_events']}")

# Top 10 registros por puntuación
print("\n=== TOP 10 REGISTROS POR PUNTUACIÓN TOTAL ===")
df_score.select(
    "sensor_id", "score_total", "avg_value", "avg_temp", "avg_humidity", "num_events"
).orderBy(col("score_total").desc()).show(10, truncate=False)

# Puntuación promedio por sensor
print("\n=== PUNTUACIÓN PROMEDIO POR SENSOR ===")
score_por_sensor = df_score.groupBy("sensor_id").agg(
    F.avg("score_total").alias("score_promedio"),
    F.max("score_total").alias("score_maximo"),
    F.min("score_total").alias("score_minimo"),
    F.stddev("score_total").alias("desviacion_score")
).orderBy(col("score_promedio").desc())

score_por_sensor.show(truncate=False)

# Sensor con mejor puntuación promedio
mejor_sensor = score_por_sensor.first()
print(f"\nSensor con mejor puntuación promedio: {mejor_sensor['sensor_id']}")
print(f"Puntuación promedio: {mejor_sensor['score_promedio']:.2f}")
print(f"Puntuación máxima alcanzada: {mejor_sensor['score_maximo']:.2f}")
````

Crea dos gráficos, convirtiendo el DF de Spark a pandas y utiliza la librería matplotlib para generar dos gráficos sobre alguna de las variables.

````python
# Convertir DataFrame de Spark a Pandas para visualización
# Seleccionamos solo las columnas necesarias para optimizar memoria
df_pandas = df_agg.select(
    "sensor_id", "avg_value", "avg_temp", "avg_humidity", "num_events"
).toPandas()

print(f"DataFrame convertido a Pandas. Shape: {df_pandas.shape}")
print(f"\nPrimeras filas:")
print(df_pandas.head())
````

````python
# GRÁFICO 1: Distribución de valores promedio por sensor (Box Plot)
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(12, 6))

# Preparar datos para box plot
sensores = df_pandas['sensor_id'].unique()
datos_boxplot = [df_pandas[df_pandas['sensor_id'] == sensor]['avg_value'].values for sensor in sensores]

plt.boxplot(datos_boxplot, labels=sensores, patch_artist=True)
plt.title('Distribución de Valores Promedio por Sensor', fontsize=16, fontweight='bold')
plt.xlabel('Sensor ID', fontsize=12)
plt.ylabel('Valor Promedio', fontsize=12)
plt.grid(axis='y', alpha=0.3, linestyle='--')
plt.tight_layout()
plt.show()

# Estadísticas del gráfico
print("\n=== ESTADÍSTICAS DEL GRÁFICO 1 ===")
for i, sensor in enumerate(sensores):
    datos = datos_boxplot[i]
    print(f"{sensor}: Media={np.mean(datos):.2f}, Mediana={np.median(datos):.2f}, Std={np.std(datos):.2f}")
````

````python
# GRÁFICO 2: Relación entre Temperatura y Humedad por Sensor (Scatter Plot)
plt.figure(figsize=(12, 8))

# Colores para cada sensor
colores = {'S1': 'red', 'S2': 'blue', 'S3': 'green', 'S4': 'orange'}

# Crear scatter plot por cada sensor
for sensor in df_pandas['sensor_id'].unique():
    datos_sensor = df_pandas[df_pandas['sensor_id'] == sensor]
    plt.scatter(
        datos_sensor['avg_temp'], 
        datos_sensor['avg_humidity'],
        c=colores.get(sensor, 'gray'),
        label=sensor,
        alpha=0.6,
        s=50
    )

plt.title('Relación entre Temperatura y Humedad por Sensor', fontsize=16, fontweight='bold')
plt.xlabel('Temperatura Promedio (°C)', fontsize=12)
plt.ylabel('Humedad Promedio (%)', fontsize=12)
plt.legend(title='Sensor', fontsize=10)
plt.grid(True, alpha=0.3, linestyle='--')
plt.tight_layout()
plt.show()

# Correlación
print("\n=== CORRELACIÓN TEMPERATURA-HUMEDAD POR SENSOR ===")
for sensor in df_pandas['sensor_id'].unique():
    datos_sensor = df_pandas[df_pandas['sensor_id'] == sensor]
    correlacion = datos_sensor['avg_temp'].corr(datos_sensor['avg_humidity'])
    print(f"{sensor}: Correlación = {correlacion:.3f}")
````

````python
# BONUS: Gráfico 3 - Serie temporal de valores promedio
# Necesitamos el timestamp para este gráfico
df_temporal_pandas = df_agg.select(
    "sensor_id", "window", "avg_value", "avg_temp", "avg_humidity"
).toPandas()

# Extraer el inicio de la ventana como timestamp
df_temporal_pandas['timestamp'] = pd.to_datetime(df_temporal_pandas['window'].apply(lambda x: x['start']))
df_temporal_pandas = df_temporal_pandas.sort_values('timestamp')

plt.figure(figsize=(14, 6))

for sensor in df_temporal_pandas['sensor_id'].unique():
    datos_sensor = df_temporal_pandas[df_temporal_pandas['sensor_id'] == sensor]
    plt.plot(
        datos_sensor['timestamp'],
        datos_sensor['avg_value'],
        marker='o',
        label=sensor,
        alpha=0.7,
        linewidth=2,
        markersize=3
    )

plt.title('Evolución Temporal de Valores Promedio por Sensor', fontsize=16, fontweight='bold')
plt.xlabel('Tiempo', fontsize=12)
plt.ylabel('Valor Promedio', fontsize=12)
plt.legend(title='Sensor', fontsize=10)
plt.grid(True, alpha=0.3, linestyle='--')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("\n=== GRÁFICOS GENERADOS EXITOSAMENTE ===")
````
