## Ejercicio 5

```python
# Escribe un programa que pida **una palabra** por teclado y diga si está en la lista de **palabras prohibidas** (`bad_words.txt`). 
# La comparación debe ser **insensible a mayúsculas** .

palabra = input("Dime una palabra: ")

def norm(s: str) -> str:
    return s.lower().replace(" ", "")

bad = set()
with open("bad_words.txt", encoding="utf-8") as f:
    for line in f:
        w = line.strip()
        if w:
            bad.add(norm(w))

bad_minusculas = {elemento.lower() for elemento in bad}


if palabra.lower() in bad_minusculas:
    print("NO CORRECTA")
else:
    if " " in palabra:
        print("Introduce una sola palabra (sin espacios).")
    else:
        print("CORRECTA")
```
