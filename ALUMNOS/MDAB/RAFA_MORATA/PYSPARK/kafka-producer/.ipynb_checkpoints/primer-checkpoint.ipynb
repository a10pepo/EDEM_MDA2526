```python
from pyspark.sql import Row
from pyspark.sql.types import StructType, StructField, StringType, IntegerType
from pyspark.sql.functions import *
```

```python
json_schema = StructType([
    StructField("sensor_id", StringType(), True),   # "S1", "S2"...
    StructField("value", DoubleType(), True),       # random.uniform...
    StructField("temperature", DoubleType(), True), # random.uniform...
    StructField("humidity", DoubleType(), True),    # random.uniform...
    StructField("status", StringType(), True),      # "OK", "WARN"...
    StructField("timestamp", DoubleType(), True),   # time.time() es float
    StructField("uuid", StringType(), True)         # uuid es string
])
```

```python
weather_df = weather_df_raw.select(
    # 1. Convertimos la columna binaria 'value' a String
    # 2. Usamos from_json con el esquema que creamos arriba
    from_json(col("value").cast("string"), json_schema).alias("data")
).select("data.*") # Sacamos todas las columnas fuera del objeto 'data'
```

```python
df_stats = weather_df.groupBy("sensor_id").agg(
    avg("temperature").alias("media_temp"),
    max("temperature").alias("max_temp"),
    count("*").alias("num_eventos")
)
```

```python
print("Iniciando nuevo stream...")
query_stats = df_stats.writeStream \
    .queryName("tabla_estadisticas") \
    .format("memory") \
    .outputMode("complete") \
    .start()
```

```python
data_info = [("S1", "Interior", "Zona A"), ("S2", "Exterior", "Zona B"), 
             ("S3", "Interior", "Zona C"), ("S4", "Exterior", "Zona A")]
df_static = spark.createDataFrame(data_info, ["sensor_id", "ubicacion", "zona"])

df_joined = weather_df.join(df_static, "sensor_id")

query = df_joined.writeStream.queryName("tabla_join").format("memory").outputMode("append").start()
time.sleep(5)
```

```python
query.stop()
```

```python
df_time = weather_df.withColumn("fecha", col("timestamp").cast("timestamp")) \
    .groupBy(window("fecha", "1 minute")) \
    .agg(
        count("*").alias("num_eventos"),          # ¿Cuántos eventos hay?
        avg("humidity").alias("humedad_media")    # ¿Cuál es la humedad media?
    ).select(
        col("window.start").alias("inicio_ventana"), 
        "num_eventos", 
        "humedad_media"
    )
```

```python
query = df_time.writeStream.queryName("tabla_tiempo").format("memory").outputMode("complete").start()
time.sleep(5)
```

```python
spark.sql("SELECT * FROM tabla_tiempo ORDER BY inicio_ventana").show(truncate=False)
```

```python
query.stop()
```

```python
df_repart = weather_df.repartition("sensor_id") \
    .withColumn("id_particion", spark_partition_id())
```

```python
query = df_distribucion.writeStream \
    .queryName("tabla_particiones") \
    .format("memory") \
    .outputMode("complete") \
    .start()

time.sleep(10)
```

```python
query = weather_df.writeStream.queryName("tabla_anomalias").format("memory").outputMode("append").start()
time.sleep(5)
```

```python
spark.sql("""
    SELECT count(*) as total_anomalos
    FROM tabla_anomalias
    WHERE temperature > (SELECT avg(temperature) FROM tabla_anomalias)
""").show()
```

```python
spark.sql("""
    SELECT sensor_id, count(*) as cant
    FROM tabla_anomalias
    WHERE temperature > (SELECT avg(temperature) FROM tabla_anomalias)
    GROUP BY sensor_id
    ORDER BY cant DESC
    LIMIT 1
""").show()
```

```python
query.stop()
```

```python
df_score = weather_df.withColumn("puntuacion", (col("value") * 0.4) + (col("temperature") * 0.4) + (col("humidity") * 0.2))
```

```python
query = df_score.writeStream.queryName("tabla_score").format("memory").outputMode("append").start()
time.sleep(5)
```

```python
spark.sql("SELECT sensor_id, puntuacion FROM tabla_score ORDER BY puntuacion DESC LIMIT 1").show()
```

```python
import time
import matplotlib.pyplot as plt

for s in spark.streams.active:
    s.stop()

query_plot = weather_df.writeStream \
    .queryName("tabla_para_graficos") \
    .format("memory") \
    .outputMode("append") \
    .start()

time.sleep(15)

df_static = spark.sql("SELECT * FROM tabla_para_graficos")
pdf = df_static.toPandas()

if not pdf.empty:
    plt.figure(figsize=(14, 6))

    plt.subplot(1, 2, 1)
    conteo_sensores = pdf['sensor_id'].value_counts()
    conteo_sensores.plot(kind='bar', color='skyblue', edgecolor='black')
    plt.title('Cantidad de Eventos por Sensor')
    plt.xlabel('Sensor ID')
    plt.ylabel('Número de Registros')
    plt.grid(axis='y', alpha=0.3)

    plt.subplot(1, 2, 2)
    plt.scatter(pdf['temperature'], pdf['humidity'], color='salmon', alpha=0.7)
    plt.title('Relación: Temperatura vs. Humedad')
    plt.xlabel('Temperatura')
    plt.ylabel('Humedad')
    plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()
else:
    print("No han llegado datos suficientes.")

query_plot.stop()
```
