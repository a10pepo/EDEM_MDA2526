```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
import plotly.express as px
from sklearn.preprocessing import MinMaxScaler
import plotly.io as pio

```

```python
df = pd.read_csv(r"C:\dev\Machine_Learning\MachineLearning_Clustering\Entregable.csv")
```

Hacer el PCA o TSME o UMAP para pasar de 15 caracteristicas a 2


```python
df['explicit'] = df['explicit'].astype(int)


numerical_cols = ['popularity', 'duration_ms', 'explicit', 'danceability', 'energy', 'key', 'loudness', 'mode', 'speechiness', 'acousticness', 'instrumentalness', 'liveness', 'valence', 'tempo', 'time_signature']

X = df[numerical_cols]


scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

df_pca = pd.DataFrame(X_pca, columns=['PC1', 'PC2'])
```

Hacer un plot de esos datos


```python
fig = px.scatter(df_pca, x='PC1', y='PC2', title='PCA - Reducción a 2 dimensiones')
fig.show()
```

Volver al dataFrame original


```python

X_clustering = X_scaled
```

Hacer un ELBOW,SHILHOTTE

```python

inertias = []
k_range = range(2, 11)
for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_clustering)
    inertias.append(kmeans.inertia_)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.plot(k_range, inertias, marker='o')
plt.xlabel('Número de clusters')
plt.ylabel('Inertia')
plt.title('Método del Codo')

# Silhouette Score
sil_scores = []
for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_clustering)
    sil_score = silhouette_score(X_clustering, labels)
    sil_scores.append(sil_score)

plt.subplot(1, 2, 2)
plt.plot(k_range, sil_scores, marker='o')
plt.xlabel('Número de clusters')
plt.ylabel('Silhouette Score')
plt.title('Análisis de Silhouette')

plt.tight_layout()
plt.show()
```
