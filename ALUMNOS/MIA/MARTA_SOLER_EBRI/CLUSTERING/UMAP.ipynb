```python
import umap
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_num = pd.read_csv('genres_clean_num.csv')
df_hot = pd.read_csv('genres_clean_comb1.csv')
df_label = pd.read_csv('genres_clean_comb2.csv')
```

```python
umap_reducer = umap.UMAP(n_components=2, random_state=42)
df_umap_num = umap_reducer.fit_transform(df_num)
df_umap_hot = umap_reducer.fit_transform(df_hot)
df_umap_label = umap_reducer.fit_transform(df_label)
```

```python
df_plot_num = pd.DataFrame({'UMAP_1': df_umap_num[:, 0], 'UMAP_2': df_umap_num[:, 1]})
df_plot_hot = pd.DataFrame({'UMAP_1': df_umap_hot[:, 0], 'UMAP_2': df_umap_hot[:, 1]})
df_plot_label = pd.DataFrame({'UMAP_1': df_umap_label[:, 0], 'UMAP_2': df_umap_label[:, 1]})

plt.figure(figsize=(18, 5))

plt.subplot(1, 3, 1)
plt.scatter(df_plot_num['UMAP_1'], df_plot_num['UMAP_2'], s=8)
plt.title('UMAP - Numérico')
plt.xlabel('UMAP 1')
plt.ylabel('UMAP 2')

plt.subplot(1, 3, 2)
plt.scatter(df_plot_hot['UMAP_1'], df_plot_hot['UMAP_2'], s=8)
plt.title('UMAP - One Hot Encoding')
plt.xlabel('UMAP 1')
plt.ylabel('UMAP 2')

plt.subplot(1, 3, 3)
plt.scatter(df_plot_label['UMAP_1'], df_plot_label['UMAP_2'], s=8)
plt.title('UMAP - Label Encoder')
plt.xlabel('UMAP 1')
plt.ylabel('UMAP 2')

plt.tight_layout()
plt.show()
```
