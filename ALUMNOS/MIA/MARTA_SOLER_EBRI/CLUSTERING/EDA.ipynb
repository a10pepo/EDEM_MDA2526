Vemos que la mitad son nulos. Podemos hacer varias cosas, eliminar las tres columnas o imputarlos. Voy a probar a borrar las columnas de 'song_name' y 'Unnamed: 0' y en 'title' reemplazar los nulos por una nueva categoría que sea 'unnamed'. También voy a eliminar las categorías 'id', 'uri', 'track_href', 'analysis_url' ya que no aportan al análisis. 

Por otro lado voy a escalar las variables numéricas usando la normalización Z-score, voy guardar un .csv que solo contenga las variables numéricas, otro que contenga las numéricas y las categóricas y otro que contenga ambas pero con las categóricas procesadas.

Voy a usar por un lado, One Hot Encoding y Laber Encoder, y guardarlo en dos csv distintos.

```python
df.drop(columns=['Unnamed: 0', 'song_name', 'id', 'uri', 'track_href', 'analysis_url'], inplace=True)
df['title']=df['title'].fillna('unnamed')
df.reset_index(drop=True, inplace=True)
```

```python
df_cat = df.select_dtypes(include='object')
df_num = df.select_dtypes(include='number')
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
df_num_scaled = sc.fit_transform(df_num)
df_num = pd.DataFrame(df_num_scaled, columns=df_num.columns)
df_num.to_csv("genres_clean_num.csv", index=False)
```

```python
df_comb = pd.concat([df_num, df_cat], axis=1)
df_comb.to_csv("genres_clean_comb.csv", index=False)
```

```python
from sklearn.preprocessing import LabelEncoder
df_cat1 = pd.get_dummies(df_cat, columns=df_cat.columns, drop_first=False)
encoders = {}
for col in df_cat.columns:
    le = LabelEncoder()
    df_cat[col] = le.fit_transform(df_cat[col])
    encoders[col] = le
    df_cat2 = df_cat.copy()

df_comb1 = pd.concat([df_num, df_cat1], axis=1)
df_comb2 = pd.concat([df_num, df_cat2], axis=1)
df_comb1.to_csv("genres_clean_comb1.csv", index=False)
df_comb2.to_csv("genres_clean_comb2.csv", index=False)
```
