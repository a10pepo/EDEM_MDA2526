Configurar KSQL

Vamos a decirle a KSQL: "Vigila el topic de las pizzas y si ves un VIP, cópialo a otro lado


Comando: 
    docker-compose exec ksql-cli ksql http://ksql:8088

1. Copio y pego este comando en terminal (Define el Stream sobre el topic que llena el Jefe de Cocina):

CREATE STREAM stream_pizzas (
    id INTEGER,
    mesa INTEGER,
    plato STRUCT<id INTEGER, tipo VARCHAR, nombre VARCHAR>,
    tipo_cliente VARCHAR
) WITH (
    KAFKA_TOPIC='pedidos_pizza',
    VALUE_FORMAT='JSON'
);

2. Copio y pego este comando en terminal. Crea el topic final solo para el Gerente:

CREATE STREAM stream_vip_final WITH (
    KAFKA_TOPIC='pizzas_vip_topic',
    VALUE_FORMAT='JSON'
) AS
SELECT * FROM stream_pizzas 
WHERE tipo_cliente = 'VIP';


3. Ahora KSQL está trabajando en segundo plano filtrando los datos.
