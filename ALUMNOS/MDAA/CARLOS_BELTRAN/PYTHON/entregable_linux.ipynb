# Ejercicio 5

```python
def norm(cadena: str)->str:
    return cadena.lower().strip()

palabraUsuario = norm(input("Escribe una palabra"))

if " " in palabraUsuario or palabraUsuario == "":
    print("Introduce una sola palabra (sin espacios).")
else:
    bad = set()
    with open("bad_words.txt", encoding="utf-8") as f:
        for line in f:
            w = line.strip()
            if w:
                bad.add(norm(w))

    if palabraUsuario in bad:
        print(f"NO CORRECTA")
    else:
        print("CORRECTA")
```
